1.Scatter Plot

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
y1=np.sin(x)
y2=np.cos(x)
```

```python
plt.figure(figsize=(8,4))
plt.scatter(x,y1,label='sin(x)',color='red')
plt.scatter(x,y2,label='cos(x)',color='blue')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Scatter Plot')
plt.legend()
plt.show()
```

2.Bar Plot

```python
categories=['A','B','C','D']
values=[10,25,15,30]
plt.figure(figsize=(8,4))
plt.bar(categories,values)
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Plot')
plt.show()
```

3.Histogram

```python
data = np.random.randn(1000)
plt.figure(figsize=(8,4))
plt.hist(data,bins=30)
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

4.Pie Chart

```python
labels = ['A','B','C']
sizes = [30,45,25]
plt.figure(figsize=(6,6))
plt.pie(sizes,labels=labels,autopct='%1.1f%%',startangle=90)
plt.title('Pie chart')
plt.show()
```

5.Line Plot

```python
x = np.linspace(0,10,100)
y1= np.sin(x)
y2= np.cos(x)
plt.figure(figsize=(8,4))
plt.plot(x,y1,label='sin(x)')
plt.plot(x,y1,label='cos(x)')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('line plot')
plt.legend()
plt.show()
```

6. Subplots

```python
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(x,y1)
plt.plot('Subplot 1')
plt.subplot(1,2,2)
plt.scatter(x,y2)
plt.title('Subplot 2')
plt.show()
```
